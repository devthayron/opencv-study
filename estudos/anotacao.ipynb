# Importação

```python
import cv2 as cv
```

## Dimensões da imagem em pixels (img.shapes)

- OBS. Necessário mencionar o caminho da imagem, nesse exemplo temos a imagem "farol.jpg".
- Resultado (altura, largura, canais de cores)

```python
IMG_PATH = "../media/farol.jpg"

img = cv.imread(IMG_PATH)

print(f"Dimensões da imagem em pixels RGB: {img.shape}") 
# print(f"Tamanho em pixels da imagem {img}")
```

## Transformando em escala de Cinza (0,255)

```python
img_gray = cv.cvtColor(img, cv.COLOR_RGB2GRAY)
print(f"Dimensões da imagem em escala de cinza: {img_gray.shape}")
```

## Exibindo a Imagem 

```python
cv.imshow("imagem_farol", img)      # quiser ver a imagem cinza muda o "img" para "img_gray"
cv.waitKey(0)  # espera até que uma tecla seja pressionada
```

## Exibindo Video


```python
VIDEO_PATH = "../media/runners.mp4"

video = cv.VideoCapture(VIDEO_PATH)

while True:
    check, img = video.read()
    # print(img.shape)
    img_redimencionada = cv.resize(img, (640, 420))  # redimencionar

    cv.imshow("video", img_redimencionada)
    if cv.waitKey(15) & 0xFF == ord("q"):
        break

print("Video encerrado!")
video.release()
cv.destroyAllWindows()
```
